# Scaled Dot-Product Attention

This notebook demonstrates the core operations in scaled dot-product attention using NumPy. It is a small educational example, not a full Transformer implementation.

```python
import numpy as np

np.set_printoptions(precision=3, suppress=True)
```

## 1. Create queries, keys, and values

For simplicity, use three input positions and a small feature dimension.

```python
rng = np.random.default_rng(7)
sequence_length = 3
d_k = 4
d_v = 4

Q = rng.normal(size=(sequence_length, d_k))
K = rng.normal(size=(sequence_length, d_k))
V = rng.normal(size=(sequence_length, d_v))

Q.shape, K.shape, V.shape
```

## 2. Calculate attention scores

The query-key dot products measure compatibility. Divide by the square root of the key dimension to scale the scores.

```python
scores = Q @ K.T / np.sqrt(d_k)
scores
```

## 3. Convert scores into weights

Softmax is applied row-wise so each query's attention weights sum to one.

```python
def softmax(x, axis=-1):
    x = x - np.max(x, axis=axis, keepdims=True)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=axis, keepdims=True)

weights = softmax(scores, axis=-1)
weights, weights.sum(axis=-1)
```

## 4. Combine the value vectors

Each output is a weighted sum of the value vectors.

```python
output = weights @ V
output
```

## What to notice

- Each row of the attention matrix corresponds to one query position.
- Each row of weights sums to one.
- The output at each position is a weighted combination of all value vectors.

Further experiments: change the random seed, inspect the weight matrix, or add a causal mask so a position cannot attend to future positions.
